```python
import numpy as np
arr_1d = np.array([1,2,3,4,5])
print(arr_1d)
```

```python
arr_2d = np.array([[2,3,4,5],[6,7,8,9]])
print(arr_2d)
```

```python
# shape - find the dimension of the array
# shape return a tuple (row,column0 ) for  2 d arrays
print("shape of arr_2d:",arr_2d.shape)
```

```python
arr_3d = np.array([[[1,2,3],[4,5,6],[8,9,7]]])
print(arr_3d)
```

```python
print("shape of 3_arr:",arr_3d.shape)
```

```python
# find the array of  data type of array element:
print("data type of arr_1d:",arr_1d.dtype)
```

```python
float_arr_1d = arr_1d.astype(float)
print("array after converting to integers:",float_arr_1d)
```

```python
print("Data type after conversion:",float_arr_1d.dtype)
```

```python
# ndim : it calculate number of dimentsions
# ndim gives how many dimesions (levels) the array has
print("number of dimension in arr_1d:",arr_1d.ndim)
```

```python
print("number of dimension in arr_2d:",arr_2d.ndim)
```

```python
print("number of dimesion in arr_3d",arr_3d.ndim)
```
